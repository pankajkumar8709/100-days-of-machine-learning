```python
import kagglehub
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import os
```

```python
path=kagglehub.dataset_download("nilesh2042/nirf-indian-university-rankings-by-gov-of-india")
path
```

```python
os.listdir(path)
```

```python
df=pd.read_csv(path+"/Institute_Rankings.csv")
```

```python
df.sample(5)
```

```python
df.shape
```

```python
df.info()
```

```python
X = df[['City', 'State', 'Score']]
y = df['Rank']
```

```python
from sklearn.preprocessing import LabelEncoder, OneHotEncoder
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_squared_error
```

```python
ohe=OneHotEncoder(drop='first',sparse_output=False)
ohe_trf=ohe.fit_transform(X[['City','State']])
```

```python
ohe.get_feature_names_out()
```

```python
x_trf=pd.DataFrame(ohe_trf,columns=ohe.get_feature_names_out())
x_trf
```

```python
X_trf=np.concatenate((x_trf.values,X[['Score']].values),axis=1)
X_trf
```

```python
y
```

```python
from sklearn.model_selection import train_test_split
X_train,X_test,y_train,y_test=train_test_split(X_trf,y,test_size=0.2,random_state=2)
```

```python
lr=LinearRegression()
lr.fit(X_train,y_train)
```

```python
r2_score=lr.score(X_train,y_train)
r2_score
```

```python
from sklearn.ensemble import RandomForestRegressor
```

```python
rf=RandomForestRegressor()
rf.fit(X_train,y_train)
```

```python
r2_score=rf.score(X_train,y_train)
r2_score
```

```python
from xgboost import XGBRegressor
```

```python
xgbr=XGBRegressor()
xgbr.fit(X_train,y_train)
```

```python
r2_score=xgbr.score(X_train,y_train)
r2_score
```
